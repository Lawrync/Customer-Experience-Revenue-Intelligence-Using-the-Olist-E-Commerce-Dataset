```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
```

```python
# Reading of CSV files 
customer = pd.read_csv("olist_customers_dataset-checkpoint.CSV")
geolocation = pd.read_csv("olist_geolocation_dataset.csv")
order_items=  pd.read_csv("olist_order_items_dataset.csv")
orders= pd. read_csv("olist_orders_dataset.csv")
payment = pd.read_csv("olist_order_payments_dataset.csv")
reviews= pd.read_csv("olist_order_reviews_dataset.csv")
product= pd.read_csv("olist_products_dataset.csv")
seller = pd.read_csv("olist_sellers_dataset.csv")
category= pd.read_csv("product_category_name_translation.csv")
```

```python
# Dictionary containing dataset names and file paths
datasets = {
    "Customer": customer,
    "Geolocation": geolocation,
    "Order Items": order,
    "Orders": order,
    "Payments": payment,
    "Reviews": reviews,
    "Products": product,
    "Sellers": seller,
    "Category Translation": category
}

```

```python
# Display information for each DataFrame
for name, df in datasets.items():
    print("=" * 80)
    print(f"{name.upper()} DATASET")
    print("=" * 80)

    # Shape
    print(f"\nShape: {df.shape}")

    # Head
    print("\nFirst 5 Rows:")
    display(df.head())  # Use display() in Jupyter Notebook

    # Info
    print("\nInformation:")
    df.info()

    # Missing Values
    print("\nMissing Values:")
    print(df.isnull().sum())

    # Duplicate Rows
    print(f"\nDuplicate Rows: {df.duplicated().sum()}")

    print("\n\n")
```

# Data Findings

## Customer Dataset

. Our dataset comprises 99441 rows and 5 columns.

. Has no missing values or duplicates.

. The data has 4 objects and one integer.

. Our customer dataset has no issues.

## Geolocation Dataset

. The geolocation dataset is the largest, containing 1,000,163 rows and 5 columns.

. It has no missing values.

. The dataset contains 2 float, 1 integer, and 2 object data types.

61,831 duplicate records should be reviewed before analysis to avoid biased results.

## Order Items Dataset

The order items dataset contains 112,650 rows and 7 columns.

There are no missing values or duplicate records.

The dataset consists of 4 object, 2 float, and 1 integer data types.

The shipping_limit_date column should be converted from object to datetime for time-based analysis.

## order Dataset 
dataset contained 99,441 records and 8 columns with no duplicate entries. 

Three delivery-related columns (order_approved_at, order_delivered_carrier_date, and order_delivered_customer_date) contained missing values. 

Rather than removing these records, the missing values were retained because they correspond to the lifecycle of an order. For example, cancelled or unavailable orders are not expected to have approval or delivery timestamps. 

All date columns were converted from the object data type to the datetime format to support temporal analyses such as sales trends, delivery duration, and order processing times

## Payments Dataset

The payments dataset contains 103,886 rows and 5 columns.

It has no missing values or duplicate records.

The dataset consists of 2 object, 2 integer, and 1 float data types.

The payments dataset is clean and ready for analysis.

## Reviews Dataset

The reviews dataset comprises 99,224 rows and 7 columns.

It contains 6 object data types and 1 integer data type.

There are significant missing values in review_comment_title (87,656) and review_comment_message (58,247) because many customers did not leave written reviews.

There are no duplicate records.

The review date columns should be converted to datetime during preprocessing.

## Products Dataset

The products dataset contains 32,951 rows and 9 columns.

It consists of 2 object data types and 7 float data types.

Missing values exist in the product category, product description, product photo count, and product dimensions.
There are no duplicate records.

The missing values should be handled before analysis, either through imputation or removal of affected records.

## Sellers Dataset

The sellers dataset contains 3,095 rows and 4 columns.

It has no missing values and no duplicate records.

The dataset consists of 3 object data types and 1 integer data type.

The sellers dataset is clean and requires minimal preprocessing.

# Category Translation Dataset

The category translation dataset is the smallest, containing 71 rows and 2 columns.

It has no missing values or duplicate records.

Both columns are of object data type.

The dataset is clean and will be used to translate Portuguese product categories into English when merging with the products dataset

```python
# Check duplicates in all datasets
for name, df in datasets.items():
    print(f"{name}: {df.duplicated().sum()} duplicate rows")
```

```python
for name, df in datasets.items():
    print(f"\n{name}")
    print(df.isnull().sum())
```

```python
# Remove duplicates from geolocation dataset
datasets["Geolocation"] = datasets["Geolocation"].drop_duplicates()
```

## Reviews Dataset


```python
reviews["review_comment_title"] = reviews["review_comment_title"].fillna("No Title")

reviews["review_comment_message"] = reviews["review_comment_message"].fillna("No Comment")
```

# Products Dataset

```python
product["product_category_name"] = product["product_category_name"].fillna("Unknown")
product["product_name_lenght"] = product["product_name_lenght"].fillna(
    product["product_name_lenght"].median()
)
product["product_description_lenght"] = product["product_description_lenght"].fillna(
    product["product_description_lenght"].median()
)
product["product_photos_qty"] = product["product_photos_qty"].fillna(
    product["product_photos_qty"].median()
)
```

```python
product["product_weight_g"] = product["product_weight_g"].fillna(
    product["product_weight_g"].median()
)

product["product_length_cm"] = product["product_length_cm"].fillna(
    product["product_length_cm"].median()
)

product["product_height_cm"] = product["product_height_cm"].fillna(
    product["product_height_cm"].median()
)

product["product_width_cm"] = product["product_width_cm"].fillna(
    product["product_width_cm"].median()
)
```

## Convert Data Types

```python
order["shipping_limit_date"] = pd.to_datetime(order["shipping_limit_date"])
```

```python
reviews["review_creation_date"] = pd.to_datetime(reviews["review_creation_date"])

reviews["review_answer_timestamp"] = pd.to_datetime(
    reviews["review_answer_timestamp"]
)
```

## Rename Columns

```python
product.rename(columns={
    "product_name_lenght": "product_name_length",
    "product_description_lenght": "product_description_length"
}, inplace=True)
```

```python
for name, df in datasets.items():

    print("="*60)
    print(name)

    print("Missing Values")
    print(df.isnull().sum())

    print("\nDuplicates")
    print(df.duplicated().sum())
```

```python
# Merge Customers with Orders
merged_df = orders.merge(
    customer,
    on="customer_id",
    how="left"
)

print(merged_df.shape)
```

```python
# Merge Order Items
merged_df = merged_df.merge(
    order_items,
    on="order_id",
    how="left"
)

print(merged_df.shape)
```

```python
# Merge Payments
merged_df = merged_df.merge(
    payment,
    on="order_id",
    how="left"
)

print(merged_df.shape)
```

```python
# Merge Reviews
merged_df = merged_df.merge(
    reviews,
    on="order_id",
    how="left"
)

print(merged_df.shape)
```

```python
# Merge Products
merged_df = merged_df.merge(
    product,
    on="product_id",
    how="left"
)

print(merged_df.shape)
```

```python
# Merge Sellers
merged_df = merged_df.merge(
    seller,
    on="seller_id",
    how="left"
)

print(merged_df.shape)
```

```python
# Merge Category Translation
merged_df = merged_df.merge(
    category,
    on="product_category_name",
    how="left"
)

print(merged_df.shape)
```

```python
merged_df.to_csv("olist_merged_dataset.csv", index=False)
```
